# SKEWNESS AND KURTOSIS

## $ Skewness = \frac{1}{N} \sum_{n = 1}^{N} (\frac{y_n - \mu}{\sigma})^3 $

## $ Kurtosis = \frac{1}{N} \sum_{n = 1}^{N} (\frac{y_n - \mu}{\sigma})^4 - 3 $

```python
from scipy.stats import skew, kurtosis
import numpy as np
import librosa
```

```python
def skew_Kurt(audio):

    skewness_ = skew(audio)
    kurtosis_ = kurtosis(audio)

    return float(skewness_), float(kurtosis_)
```

```python
sr = 8000
file =  'test_pathological_voice.wav'
audio, sr = librosa.load(file, sr=sr, mono=True)
audio
```

```python
skewness, kurtos = skew_Kurt(audio)
print(f'Skew = {skewness}')
print(f'Kurtosis = {kurtos}')
```
